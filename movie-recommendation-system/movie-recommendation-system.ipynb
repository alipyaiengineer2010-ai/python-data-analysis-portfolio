```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df = pd.read_csv("movie.csv")
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
df['About']=df['About'].fillna("")
```

```python
tf = TfidfVectorizer()

about_int = tf.fit_transform(df['About'])
```

```python
user = input("چه فیلمی دوست داری ؟")
user_int = tf.transform([user])
```

```python
shebahat = cosine_similarity(user_int,about_int).flatten()
```

```python
top_index = shebahat.argsort()[-5:][::-1]
```

```python
for i in top_index:
    print(f"فیلم برتر {df.iloc[i]['Name']}")
    print(f" امتیاز فیلم {df.iloc[i]['Score']}")
    print(f" توضیحات: {df.iloc[i]['About']}")
    print(f" ژانر: {df.iloc[i]['Genre']}")

    print("-"*50)
```
